```python
import pandas as pd 
from pathlib import Path
import matplotlib.pyplot as plt
from util import ROOT
```

```python
csv_paths = (ROOT / "rf_models").glob("results*.csv")

for csv_path in csv_paths:
    ncl_cls = csv_path.stem.split("_")[-1]
    df = pd.read_csv(csv_path)

    plt.figure()
    plt.plot(df["thresholds"], df["precision"])
    plt.title(f"Precision - class {ncl_cls}")
    plt.xlabel("Threshold")
    plt.ylabel("Precision")
    plt.ylim(0, 1)
    plt.show()

    plt.figure()
    plt.plot(df["thresholds"], df["recall"])
    plt.title(f"Recall - class {ncl_cls}")
    plt.xlabel("Threshold")
    plt.ylabel("Recall")
    plt.ylim(0, 1)
    plt.show()

    plt.figure()
    plt.plot(df["thresholds"], df["f1"])
    plt.title(f"F1 - class {ncl_cls}")
    plt.xlabel("Threshold")
    plt.ylabel("F1")
    plt.ylim(0, 1)
    plt.show()
```
